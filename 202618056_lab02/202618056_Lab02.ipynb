## The Difference: np.linspace() vs. np.arange()

np.linspace() is driven by the number of elements you want. You specify the exact start value, the end value, and the total count. NumPy handles the math to space them perfectly. By default, it includes the final stop value.

np.arange() is driven by the step size between elements. You specify the start value, the end value, and how much to count by (the increment step). By default, it excludes the final stop value.

## Task 2 - Vectorized Arithmetic and Linear Algebra

● Create two matrices. Perform addition, element-wise multiplication, and matrix multiplication using @
or np.matmul().

● For a square matrix, find its transpose and determinant.

● If invertible, find the inverse and verify A @ A^-1 using np.allclose().

```python
import numpy as np

# Create two 2x2 matrices
matrix_A = np.array([[1, 2], [3, 4]])
matrix_B = np.array([[5, 6], [7, 8]])

print("=== Matrix A ===")
print(matrix_A)

print("\n=== Matrix B ===")
print(matrix_B)

# 1. Matrix Addition (adds matching positions)
addition_result = matrix_A + matrix_B
print("\n--- 1. Matrix Addition (A + B) ---")
print(addition_result)

# 2. Element-Wise Multiplication (multiplies matching positions)
element_mult_result = matrix_A * matrix_B
print("\n--- 2. Element-wise Multiplication (A * B) ---")
print(element_mult_result)

# 3. Matrix Multiplication / Dot Product (Rows times Columns)
# Using the '@' operator
matrix_mult_result = matrix_A @ matrix_B
print("\n--- 3. Matrix Multiplication (A @ B) ---")
print(matrix_mult_result)
```

```python
import numpy as np

# Create a square 2x2 matrix
matrix_square = np.array([[1, 2], [3, 4]])

print("=== Original Square Matrix ===")
print(matrix_square)

# 1. Find the Transpose (Flips rows into columns)
# You can use .T or np.transpose()
matrix_transpose = matrix_square.T

print("\n--- 1. Transposed Matrix (.T) ---")
print(matrix_transpose)

# 2. Find the Determinant 
# Uses the linear algebra submodule (linalg)
determinant = np.linalg.det(matrix_square)

print("\n--- 2. Determinant ---")
print(f"Determinant value: {determinant}")
```

```python
import numpy as np

# Define our square matrix A (from the previous step)
matrix_A = np.array([[1, 2], [3, 4]])

# Step 1: Calculate the determinant to check if it is invertible
det = np.linalg.det(matrix_A)
print(f"Determinant of A: {det}")

# Check if the determinant is non-zero (invertible)
if not np.isclose(det, 0):
    # Step 2: Find the inverse matrix (A^-1)
    matrix_inverse = np.linalg.inv(matrix_A)
    print("\n--- 1. Inverse Matrix (A^-1) ---")
    print(matrix_inverse)
    
    # Step 3: Verify that A @ A^-1 equals the Identity Matrix
    identity_check = matrix_A @ matrix_inverse
    print("\n--- 2. Result of A @ A^-1 ---")
    print(identity_check)
    
    # Generate an actual Identity Matrix to compare against
    expected_identity = np.eye(matrix_A.shape[0])
    
    # Step 4: Use np.allclose() to handle tiny floating-point rounding issues
    is_verified = np.allclose(identity_check, expected_identity)
    print(f"\nVerification successful (Matches Identity Matrix)? {is_verified}")

else:
    print("The matrix is singular and cannot be inverted because its determinant is 0.")
```

```python
import numpy as np

# =====================================================================
# 1. SETUP AND ELEMENT-WISE ARITHMETIC
# =====================================================================
# Create two distinct 2x2 square matrices
matrix_A = np.array([[2, 4], 
                     [1, 7]])

matrix_B = np.array([[5, 1], 
                     [3, 2]])

print("=== Matrix A ===")
print(matrix_A)

print("\n=== Matrix B ===")
print(matrix_B)

# Addition (Vectorized)
print("\n--- Matrix Addition (A + B) ---")
print(matrix_A + matrix_B)

# Element-wise Multiplication (Vectorized)
print("\n--- Element-wise Multiplication (A * B) ---")
print(matrix_A * matrix_B)


# =====================================================================
# 2. LINEAR ALGEBRA MATRIX MULTIPLICATION
# =====================================================================
# Dot product multiplication using the '@' operator
matrix_multiplication = matrix_A @ matrix_B

print("\n--- Matrix Multiplication (A @ B) ---")
print(matrix_multiplication)


# =====================================================================
# 3. TRANSPOSE AND DETERMINANT
# =====================================================================
# Transpose of Matrix A
matrix_A_transpose = matrix_A.T
print("\n--- Transpose of Matrix A (.T) ---")
print(matrix_A_transpose)

# Determinant of Matrix A
det_A = np.linalg.det(matrix_A)
print(f"\nDeterminant of Matrix A: {det_A:.4f}")


# =====================================================================
# 4. INVERSE MATRIX VERIFICATION
# =====================================================================
# Check if invertible (determinant must not be 0)
if not np.isclose(det_A, 0):
    # Calculate the Inverse (A^-1)
    matrix_A_inverse = np.linalg.inv(matrix_A)
    print("\n--- Inverse of Matrix A (A^-1) ---")
    print(matrix_A_inverse)
    
    # Calculate A @ A^-1
    product_test = matrix_A @ matrix_A_inverse
    print("\n--- Result of A @ A^-1 ---")
    print(product_test)
    
    # Generate a matching Identity Matrix (1s on diagonal, 0s elsewhere)
    identity_matrix = np.eye(matrix_A.shape[0])
    
    # Verify using np.allclose to handle microscopic floating-point rounding
    verification = np.allclose(product_test, identity_matrix)
    print(f"\nDoes A @ A^-1 equal the Identity Matrix? {verification}")
else:
    print("\nMatrix A is singular and cannot be inverted.")
```

## Task 3 - Normal Distribution and Histogram

● Generate at least 1,000 values from a normal distribution. State the chosen mean and standard
deviation.

● Find the sample mean and sample standard deviation. Compare them with the chosen values.

● Plot a histogram with it.

```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================================
# STEP 1: SET PARAMETERS & GENERATE NORMAL DISTRIBUTION DATA
# =====================================================================
# Set random seed for reproducibility
rng = np.random.default_rng(seed=42)

# Define chosen parameters (Population Parameters)
chosen_mean = 50.0
chosen_std = 10.0
num_samples = 1000

# Generate 1,000 values from the normal distribution
data = rng.normal(loc=chosen_mean, scale=chosen_std, size=num_samples)

print(f"--- Population Configuration ---")
print(f"Chosen Mean: {chosen_mean}")
print(f"Chosen Standard Deviation: {chosen_std}")
print(f"Sample Size: {num_samples}\n")


# =====================================================================
# STEP 2: CALCULATE AND COMPARE SAMPLE STATISTICS
# =====================================================================
# Find the actual sample mean and standard deviation from the generated data
sample_mean = np.mean(data)
sample_std = np.std(data)

# Calculate the differences
mean_diff = abs(chosen_mean - sample_mean)
std_diff = abs(chosen_std - sample_std)

print(f"--- Calculated Sample Statistics ---")
print(f"Sample Mean: {sample_mean:.4f} (Difference from chosen: {mean_diff:.4f})")
print(f"Sample Std Dev: {sample_std:.4f} (Difference from chosen: {std_diff:.4f})")


# =====================================================================
# STEP 3: PLOT THE HISTOGRAM
# =====================================================================
plt.figure(figsize=(8, 5))

# Plot the distribution histogram with 30 bins
plt.hist(data, bins=30, edgecolor='black', color='skyblue', alpha=0.7, density=True)

# Add a vertical line marking the calculated Sample Mean
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=2, label=f'Sample Mean ({sample_mean:.2f})')

# Add titles, labels, and formatting
plt.title(f"Normal Distribution Histogram (N = {num_samples})", fontsize=14)
plt.xlabel("Values", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.legend()
plt.grid(axis='y', alpha=0.3)

# Display the final plot window
plt.show()
```
